# Preparing the data

```python
import numpy as np              # numerical operations and array handling
import random                   # generating random numbers or sampling
import matplotlib.pyplot as plt # plotting/visualization: create figures, scatter/line plots, legends, styles
from sklearn.linear_model import Perceptron          # simple linear classifier (binary or multiclass)
from sklearn.linear_model import LogisticRegression  # statistical model estimating probabilities for classification (sigmoid-based)
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
from sklearn.datasets import fetch_openml 
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import GridSearchCV
import time
```

```python
#importing the MNIST dataset
#naming the data as X for the images, and y for the label

mnist = fetch_openml('mnist_784', version=1)
X = mnist.data
y = mnist.target
```

## Splitting the data into training and test set

```python
X_train = X[:60000]
y_train = y[:60000]
X_test = X[60000:]
y_test = y[60000:]
```

## Building the model

```python
clf = LogisticRegression(penalty=None, max_iter=1000)
```

```python
scaler = MinMaxScaler()
norm_x_train = scaler.fit_transform(X_train)
norm_x_test = scaler.transform(X_test)
```

```python
clf.fit(norm_x_train, y_train)
print("Test Accuracy:", clf.score(norm_x_test, y_test))

y_pred = clf.predict(norm_x_test)
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

print("number of iterations:", clf.n_iter_)
```

## Improving the model

### Trying different paramaters with each other

```python

c_values = [10.0, 50.0, 100.0]

for c in c_values:
    log_model = LogisticRegression(C=c, max_iter=1000, n_jobs=-1)
    
    log_model.fit(norm_x_train, y_train)
    
    score = log_model.score(norm_x_test, y_test)
    
    print(f"C = {c} | Accuracy: {score * 100:.2f}%")
```
